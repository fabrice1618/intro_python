## Variables et affectation

### Swap ★

Échanger les valeurs des variables `x` et `y` à l'aide d'une troisième variable.

Résultat attendu :

```
avant : x = 10, y = 20
après : x = 20, y = 10
```

````python
x = 10
y = 20
print(f"avant : x = {x}, y = {y}")

temporaire = x      # sauvegarder x avant de l'écraser
x = y
y = temporaire

print(f"après : x = {x}, y = {y}")
````

### Permutation circulaire ★

Faire tourner les valeurs de trois variables : la valeur de `x` passe dans `y`, celle de `y` dans `z`, et celle de `z` dans `x`.

Résultat attendu :

```
avant : x = 10, y = 20, z = 30
après : x = 30, y = 10, z = 20
```

````python
x = 10
y = 20
z = 30
print(f"avant : x = {x}, y = {y}, z = {z}")

temporaire = z      # sauvegarder z avant de l'écraser
z = y
y = x
x = temporaire

print(f"après : x = {x}, y = {y}, z = {z}")
````

### Calcul TVA ★

Ecrire un programme qui lit le prix HT d’un article, le nombre d’articles et le taux de TVA, et qui fournit le prix total TTC correspondant. Faire en sorte que des libellés apparaissent clairement.

Le prix peut avoir des centimes : le convertir avec `float()`. Afficher les totaux avec 2 décimales.

Résultat attendu :

```
Prix HT d'un article : 15.28
Nombre d'articles : 3
Taux de TVA en % : 20
```

````python
prix_ht = float(input("Prix HT d'un article : "))
nb_articles = int(input("Nombre d'articles : "))
taux_tva = float(input("Taux de TVA en % : "))

total_ht = prix_ht * nb_articles
total_ttc = total_ht * (1 + taux_tva / 100)
print(f"Total HT : {total_ht:.2f} €")
print(f"Total TTC : {total_ttc:.2f} €")
````

### Volume de la chambre froide ★

Une chambre froide a la forme d'un pavé droit. Demander sa longueur, sa largeur et sa hauteur en mètres, puis afficher :

1. son volume en m³ et sa surface au sol en m², avec 2 décimales ;
2. le nombre de palettes de 1,2 m × 0,8 m que l'on peut poser au sol, toutes dans le même sens : calculer combien en tiennent dans la longueur et dans la largeur avec la division entière `//`.

Résultat attendu :

```
Longueur (m) : 5
Largeur (m) : 3.5
Hauteur (m) : 2.5
Volume : 43.75 m³
Surface au sol : 17.50 m²
Palettes au sol : 16
```

````python
longueur = float(input("Longueur (m) : "))
largeur = float(input("Largeur (m) : "))
hauteur = float(input("Hauteur (m) : "))

volume = longueur * largeur * hauteur
surface = longueur * largeur
print(f"Volume : {volume:.2f} m³")
print(f"Surface au sol : {surface:.2f} m²")

# Palettes de 1.2 m x 0.8 m : combien en tient-il dans chaque dimension ?
palettes_longueur = int(longueur // 1.2)   # // entre flottants donne un flottant : int() le convertit
palettes_largeur = int(largeur // 0.8)
print("Palettes au sol :", palettes_longueur * palettes_largeur)
````

### Conversion d'une durée ★★

Demander une durée en secondes et l'afficher en heures, minutes et secondes. Utiliser la division entière `//` et le reste de la division `%`.

Exemple pour 3725 secondes :

```

```

````python
duree = int(input("Durée en secondes : "))
heures = duree // 3600
minutes = (duree % 3600) // 60
secondes = duree % 60
print(f"{heures} h {minutes} min {secondes} s")
````

### Étiquette de relevé ★★

Un capteur envoie ses relevés sous forme de lignes de texte. Les champs ont une **largeur fixe** : la date (10 caractères), l'heure (5), le lieu (15) et la température, séparés par des `;`.

```python
ligne = "03/11/2020;08:00;chambrefroide01;3,5"
```

1. Extraire la date, l'heure, le lieu et la température par slicing.
2. Convertir la température en nombre : remplacer la virgule par un point, puis utiliser `float()`.
3. Extraire le numéro de la chambre (les 2 derniers caractères du lieu) et le convertir en entier.
4. Écrire la date au format international `2020-11-03`.
5. Afficher l'étiquette ci-dessous. L'écart à la consigne est la température moins 4 °C.

Résultat attendu :

```
Chambre n° 1 (chambrefroide01)
Relevé du 2020-11-03 à 08:00
Température : 3.5 °C, écart à la consigne : -0.5 °C
```

````python
ligne = "03/11/2020;08:00;chambrefroide01;3,5"

# 1. Extraire les champs : ils sont à des positions fixes
date = ligne[0:10]
heure = ligne[11:16]
lieu = ligne[17:32]
texte_temperature = ligne[33:]

# 2. Convertir la température
temperature = float(texte_temperature.replace(",", "."))

# 3. Numéro de la chambre : les 2 derniers caractères du lieu
numero = int(lieu[-2:])

# 4. Date au format international : année-mois-jour
date_internationale = f"{date[6:]}-{date[3:5]}-{date[:2]}"

# 5. Étiquette
ecart = temperature - 4.0
print(f"Chambre n° {numero} ({lieu})")
print(f"Relevé du {date_internationale} à {heure}")
print(f"Température : {temperature} °C, écart à la consigne : {ecart:.1f} °C")
````

### Registre d'état ★★

Le régulateur d'une chambre froide résume son état dans un entier de 8 bits, appelé **registre**. Chaque bit donne une information :

| Bit | Masque | Signification quand le bit vaut 1 |
|---|---|---|
| 0 | `COMPRESSEUR = 1 << 0` | compresseur en marche |
| 2 | `PORTE = 1 << 2` | porte ouverte |
| 5 | `ALARME = 1 << 5` | alarme de température |
| 7 | `DEGIVRAGE = 1 << 7` | dégivrage en cours |

Le registre vaut `etat = 0b0010_0101` : le `_` sépare les chiffres pour la lisibilité, Python l'ignore.

1. Afficher le registre en binaire sur 8 bits.
2. Pour chacune des 4 informations, afficher `True` ou `False` en testant le bit avec `&`.
3. Modifier le registre et l'afficher sur 8 bits après chaque étape :
   - lancer le dégivrage : mettre le bit 7 à 1 avec `|` ;
   - acquitter l'alarme : mettre le bit 5 à 0 avec `&` et `~` ;
   - basculer le compresseur : inverser le bit 0 avec `^`.
4. Bonus : afficher le nombre de bits à 1 du registre final, avec `bin()` et la méthode `count()` des chaînes.

Résultat attendu :

```
registre : 00100101
compresseur en marche : True
porte ouverte : True
alarme : True
dégivrage : False
dégivrage lancé : 10100101
alarme acquittée : 10000101
compresseur basculé : 10000100
bits à 1 : 2
```

````python
COMPRESSEUR = 1 << 0
PORTE = 1 << 2
ALARME = 1 << 5
DEGIVRAGE = 1 << 7

etat = 0b0010_0101

# 1. Afficher le registre sur 8 bits
print(f"registre : {etat:08b}")

# 2. Tester chaque bit avec & : le résultat est non nul si le bit vaut 1
print(f"compresseur en marche : {(etat & COMPRESSEUR) != 0}")
print(f"porte ouverte : {(etat & PORTE) != 0}")
print(f"alarme : {(etat & ALARME) != 0}")
print(f"dégivrage : {(etat & DEGIVRAGE) != 0}")

# 3. Modifier le registre
etat = etat | DEGIVRAGE     # mettre le bit 7 à 1
print(f"dégivrage lancé : {etat:08b}")
etat = etat & ~ALARME       # mettre le bit 5 à 0
print(f"alarme acquittée : {etat:08b}")
etat = etat ^ COMPRESSEUR   # inverser le bit 0
print(f"compresseur basculé : {etat:08b}")

# 4. Bonus : compter les bits à 1
print("bits à 1 :", bin(etat).count("1"))
````
