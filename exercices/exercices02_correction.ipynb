### Conversion d'une durée ★★

Demander une durée en secondes et l'afficher en heures, minutes et secondes. Utiliser la division entière `//` et le reste de la division `%`.

Exemple pour 3725 secondes :

```
Durée en secondes : 3725
1 h 2 min 5 s
```

````python
duree = int(input("Durée en secondes : "))
heures = duree // 3600
minutes = (duree % 3600) // 60
secondes = duree % 60
print(f"{heures} h {minutes} min {secondes} s")
````

### Ticket de caisse ★★

Un client achète 3 cahiers à 2,35 €, 2 stylos à 1,20 € et 1 cartable à 24,90 €. Afficher le ticket ci-dessous avec des f-strings : colonnes alignées, prix avec 2 décimales, total HT puis total TTC (TVA de 20 %).

```
article     qté    prix    total
cahier        3    2.35     7.05
stylo         2    1.20     2.40
cartable      1   24.90    24.90
Total HT                   34.35
Total TTC                  41.22
```

````python
tva = 20

quantite_cahier, prix_cahier = 3, 2.35
quantite_stylo, prix_stylo = 2, 1.20
quantite_cartable, prix_cartable = 1, 24.90

total_cahier = quantite_cahier * prix_cahier
total_stylo = quantite_stylo * prix_stylo
total_cartable = quantite_cartable * prix_cartable
total_ht = total_cahier + total_stylo + total_cartable
total_ttc = total_ht * (1 + tva / 100)

print(f"{'article':<10}{'qté':>5}{'prix':>8}{'total':>9}")
print(f"{'cahier':<10}{quantite_cahier:>5}{prix_cahier:>8.2f}{total_cahier:>9.2f}")
print(f"{'stylo':<10}{quantite_stylo:>5}{prix_stylo:>8.2f}{total_stylo:>9.2f}")
print(f"{'cartable':<10}{quantite_cartable:>5}{prix_cartable:>8.2f}{total_cartable:>9.2f}")
print(f"{'Total HT':<23}{total_ht:>9.2f}")
print(f"{'Total TTC':<23}{total_ttc:>9.2f}")
````

### Triangle d'étoiles ★ à ★★

Demander un nombre de lignes `n` et afficher un triangle d'étoiles. Pour `n = 4` :

```
*
**
***
****
```

1. ★ Avec une boucle `for` et la réplication de chaîne `"*" * i`.
2. ★★ Sans la réplication : avec deux boucles imbriquées.
3. Bonus : une pyramide centrée.

````python
n = int(input("Nombre de lignes : "))

# 1. avec la réplication
for i in range(1, n + 1):
    print("*" * i)

# 2. avec deux boucles imbriquées
for i in range(1, n + 1):
    ligne = ""
    for j in range(i):
        ligne += "*"
    print(ligne)

# 3. bonus : pyramide centrée
for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
````

### Moyenne des saisies ★★

Demander des notes à l'utilisateur jusqu'à ce qu'il saisisse `fin`, puis afficher le nombre de notes, la moyenne, la note minimale et la note maximale, **sans utiliser de liste**. Une note hors de l'intervalle 0 à 20 est refusée avec un message.

Indice : `while True`, `break` et `continue`, un compteur et un accumulateur.

````python
nb_notes = 0
total = 0
note_min = 20
note_max = 0

while True:
    saisie = input("Note (ou fin) : ")
    if saisie == "fin":
        break
    note = float(saisie)
    if note < 0 or note > 20:
        print("Erreur : la note doit être comprise entre 0 et 20")
        continue
    nb_notes += 1
    total += note
    if note < note_min:
        note_min = note
    if note > note_max:
        note_max = note

if nb_notes > 0:
    print(f"{nb_notes} notes, moyenne {total / nb_notes:.2f}, min {note_min}, max {note_max}")
else:
    print("Aucune note saisie")
````

### Année bissextile ★★

Une année est bissextile si elle est divisible par 4 mais pas par 100, ou si elle est divisible par 400.

1. Écrire la fonction `est_bissextile(annee)` qui renvoie `True` ou `False`, puis exécuter la cellule de tests.
2. Afficher les années bissextiles de 1990 à 2030.

````python
def est_bissextile(annee):
    return (annee % 4 == 0 and annee % 100 != 0) or annee % 400 == 0
````

````python
# Tests : ne pas modifier
assert est_bissextile(2024)
assert not est_bissextile(2023)
assert not est_bissextile(1900)
assert est_bissextile(2000)
print("Tous les tests sont passés")
````

````python
for annee in range(1990, 2031):
    if est_bissextile(annee):
        print(annee)
````

### Menu ★★

Un compte bancaire a un solde initial de 100 €. Afficher en boucle le menu ci-dessous et traiter le choix de l'utilisateur avec `match`, jusqu'à ce qu'il choisisse `q` :

```
1 - afficher le solde
2 - déposer
3 - retirer
q - quitter
```

Un retrait supérieur au solde est refusé. Un choix inconnu affiche un message d'erreur.

````python
solde = 100.0

while True:
    print("1 - afficher le solde")
    print("2 - déposer")
    print("3 - retirer")
    print("q - quitter")
    choix = input("Votre choix : ")

    match choix:
        case "1":
            print(f"Solde : {solde:.2f} €")
        case "2":
            montant = float(input("Montant à déposer : "))
            solde += montant
        case "3":
            montant = float(input("Montant à retirer : "))
            if montant > solde:
                print("Refusé : solde insuffisant")
            else:
                solde -= montant
        case "q":
            print("Au revoir")
            break
        case _:
            print("Choix inconnu")
````

### Inventaire ★★

À partir de la liste `inventaire` (cellule suivante) :

1. afficher l'inventaire sous forme de tableau aligné ;
2. afficher la liste des articles en rupture de stock ;
3. calculer la valeur totale du stock (prix × quantité en stock) ;
4. trouver l'article le plus cher, sans utiliser `max`.

Résultat attendu :

```
article         prix  stock
clavier        25.00     12
souris         12.50      0
écran         149.90      4
câble HDMI      8.00      0
casque         39.90      7
Rupture de stock : souris, câble HDMI
Valeur du stock : 1178.90 €
Article le plus cher : écran
```

````python
inventaire = [
    {"article": "clavier", "prix": 25.0, "stock": 12},
    {"article": "souris", "prix": 12.5, "stock": 0},
    {"article": "écran", "prix": 149.9, "stock": 4},
    {"article": "câble HDMI", "prix": 8.0, "stock": 0},
    {"article": "casque", "prix": 39.9, "stock": 7},
]
````

````python
# 1. Afficher l'inventaire
print(f"{'article':<12}{'prix':>8}{'stock':>7}")
for produit in inventaire:
    print(f"{produit['article']:<12}{produit['prix']:>8.2f}{produit['stock']:>7}")

# 2. Articles en rupture de stock
rupture = []
for produit in inventaire:
    if produit["stock"] == 0:
        rupture.append(produit["article"])
print("Rupture de stock :", ", ".join(rupture))

# 3. Valeur totale du stock
valeur = 0
for produit in inventaire:
    valeur += produit["prix"] * produit["stock"]
print(f"Valeur du stock : {valeur:.2f} €")

# 4. Article le plus cher
plus_cher = inventaire[0]
for produit in inventaire:
    if produit["prix"] > plus_cher["prix"]:
        plus_cher = produit
print("Article le plus cher :", plus_cher["article"])
````

### Palindromes et anagrammes ★★

1. Un palindrome se lit de la même façon dans les deux sens, sans tenir compte des majuscules ni des espaces : `Kayak`, `Esope reste ici et se repose`. Écrire la fonction `est_palindrome(texte)`.
2. Deux mots sont des anagrammes s'ils sont formés des mêmes lettres : `chien` et `niche`. Écrire la fonction `sont_anagrammes(mot1, mot2)`.

Indice : `sorted("niche")` renvoie la liste triée des caractères de la chaîne.

````python
def nettoie(texte):
    return texte.lower().replace(" ", "")

def est_palindrome(texte):
    texte = nettoie(texte)
    return texte == texte[::-1]

def sont_anagrammes(mot1, mot2):
    return sorted(nettoie(mot1)) == sorted(nettoie(mot2))
````

````python
# Tests : ne pas modifier
assert est_palindrome("kayak")
assert est_palindrome("Esope reste ici et se repose")
assert not est_palindrome("python")
assert sont_anagrammes("chien", "niche")
assert sont_anagrammes("Marie", "aimer")
assert not sont_anagrammes("chat", "chien")
print("Tous les tests sont passés")
````

### Carnet de notes ★★★

Le dictionnaire `carnet` associe à chaque élève la liste de ses notes (cellule suivante).

1. Écrire la fonction `moyenne(notes)` qui renvoie la moyenne d'une liste de notes, ou `None` si la liste est vide.
2. Écrire la fonction `affiche_carnet(carnet)` qui affiche la moyenne de chaque élève avec 2 décimales, ou `pas de note`.
3. Afficher l'élève qui a la meilleure moyenne.
4. Ajouter la note 13 à David et la note 15 à Bruno, puis afficher de nouveau le carnet.

Résultat attendu :

```
Alice    14.00
Bruno     9.50
Chloé    17.50
David    pas de note
Meilleure moyenne : Chloé (17.50)
Alice    14.00
Bruno    10.60
Chloé    17.50
David    13.00
```

````python
carnet = {
    "Alice": [14, 16, 12],
    "Bruno": [9, 11, 8, 10],
    "Chloé": [17, 18],
    "David": [],
}
````

````python
def moyenne(notes):
    if len(notes) == 0:
        return None
    return sum(notes) / len(notes)

def affiche_carnet(carnet):
    for nom, notes in carnet.items():
        m = moyenne(notes)
        if m is None:
            print(f"{nom:<8} pas de note")
        else:
            print(f"{nom:<8} {m:5.2f}")

affiche_carnet(carnet)

meilleur = ""
meilleure_moyenne = -1
for nom, notes in carnet.items():
    m = moyenne(notes)
    if m is not None and m > meilleure_moyenne:
        meilleur = nom
        meilleure_moyenne = m
print(f"Meilleure moyenne : {meilleur} ({meilleure_moyenne:.2f})")

carnet["David"].append(13)
carnet["Bruno"].append(15)
affiche_carnet(carnet)
````
